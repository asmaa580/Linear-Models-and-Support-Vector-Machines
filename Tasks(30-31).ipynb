# (Softmax part)
Softmax is suitable for multiclass classification because it outputs class probabilities. On Digits, our best model achieved 96.39% validation accuracy and 96.11% test accuracy with eta=1.0 and lambda=0.001. On Iris, increasing regularization from lambda=0 to lambda=1 reduced validation accuracy from 93.33% to 83.33%. Unlike SVM, Softmax outputs probabilities rather than margins. It also uses logistic loss (cross entropy) in its multiclass form to penalize a prediction that assigns a very low probability to the correct class.
